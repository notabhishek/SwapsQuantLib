## Visualizing eigen values and vectors 

A = [[4, 1], [2, 3]]

To compute eigen values we do 
(A- lam . I) . v = 0, where v is the eigen vector

=> A - lam.I is not invertible other wise we multiply inverse both sides and v=0
=> det(A-lam.I) = 0, it is singular

We get lambda values 2, 5 after solving det = 0

```python
# plot vectors on the circle of radius 1 (magnitude 1)
# around origin after the matrix transformation

x_start = -1
x_end = 1
x_step = 0.00001

x = []
y = []

xi = x_start
while xi < x_end:
    yi = (1 - xi**2)**0.5
    x.append(xi)
    y.append(yi)

    x.append(xi)
    y.append(-yi)
    xi += x_step
```

## apply the matrix transformation to all the vectors
A . v

```python
import numpy as np

A = np.array([[4, 1], [2, 3]]) 

x_transformed = []
y_transformed = []

for i in range(len(x)):
    v = np.array([x[i], y[i]])
    v_transformed = np.matmul(A, v)
    x_transformed.append(v_transformed[0])
    y_transformed.append(v_transformed[1])
```

```python
import matplotlib.pyplot as plt
plt.plot(x_transformed, y_transformed, label = 'Transformed space')
plt.plot(x, y, label='Original space')

# Eigen values of the matrix = 2 and 5
# Eigen vector for lambda = 2 is [1, -2]
# Eigen vector for lambda = 5 is [1, 1]

# can you plot the
# eigen vectors of the matrix as well?
lambda_1 = 2
eigen_vector_1 = np.array([-1, 2])
lambda_2 = 5
eigen_vector_2 = np.array([1, 1])

# scale the eigen vectors for better visualization
eigen_vector_1_scaled = eigen_vector_1 * lambda_1
eigen_vector_2_scaled = eigen_vector_2 * lambda_2
plt.plot([0, eigen_vector_1_scaled[0]], [0, eigen_vector_1_scaled[1]], label='Eigen Vector 1 (lambda=2)', color='red')
plt.plot([0, eigen_vector_2_scaled[0]], [0, eigen_vector_2_scaled[1]], label='Eigen Vector 2 (lambda=5)', color='green')

plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.show()
```

### Observation
The matrix A transforms the space of vectors (circle) into an ellipse which is 

- stretched maximum along eigen_vector2 [1,1] (`y=x`) with magniude eigen_value2 = 5 
- stretched minimum along eigen_vector1 [-1, 2](`2x + y = 0`) with magnitude eigen_value1 = 2
